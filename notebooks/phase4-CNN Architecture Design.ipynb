```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, regularizers
import numpy as np

print("="*80)
print("DAY 42: CHARACTER-LEVEL BRANCH REBALANCING")
print("="*80)
print()

# Redefine paths for Day 42
PROJECT_ROOT = Path("MAJOR-PROJECT(SQL)")
PHASE4_ROOT = PROJECT_ROOT / "phase4_architecture"
DAY42_ARTIFACTS = PHASE4_ROOT / "artifacts/day42_char_branch"
DAY42_ARTIFACTS.mkdir(parents=True, exist_ok=True)

print("[INITIALIZING] Day 42 Character Branch Implementation")
print(f"Artifacts Directory: {DAY42_ARTIFACTS}")
print()

# Load character training data
print("="*80)
print("LOADING CHARACTER TRAINING DATA")
print("="*80)
print()

train_char_path = Path("../notebooks/phase3b_pipeline/data/tokenized/train_char_tokenized.parquet")
char_train_df = pd.read_parquet(train_char_path)

print(f"Loaded character training data: {len(char_train_df):,} samples")
print(f"Columns: {list(char_train_df.columns)}")
print(f"Label distribution:")
print(f"  Benign (0):   {(char_train_df['label'] == 0).sum():,}")
print(f"  Malicious (1): {(char_train_df['label'] == 1).sum():,}")
print()

# Analyze character sequence dimensions
max_char_length = char_train_df['char_length'].max()
mean_char_length = char_train_df['char_length'].mean()
median_char_length = char_train_df['char_length'].median()

print(f"Character Sequence Statistics:")
print(f"  Max Length:    {max_char_length}")
print(f"  Mean Length:   {mean_char_length:.2f}")
print(f"  Median Length: {median_char_length:.0f}")
print(f"  Input Dimension: {max_char_length}")
print()

print("="*80)
print("CHARACTER BRANCH ARCHITECTURE - REBALANCED (64 filters)")
print("="*80)
print()

# Build character branch with 64 filters (reduced from 128)
def build_character_branch(max_length=1024, vocab_size=256):
    """
    Build character-level branch with:
    - 64 filters (REDUCED - not 128)
    - 3 hierarchical Conv1D layers (kernels: 3, 5, 7)
    - Batch normalization after each conv layer
    - Dropout 0.5 (highest among branches)
    - L2 regularization 0.01 on conv weights
    
    Purpose: Reduce character branch dominance and control overfitting
    """
    
    char_input = layers.Input(shape=(max_length,), name='char_input', dtype=tf.int32)
    
    # Layer 1: Embedding
    x = layers.Embedding(
        input_dim=vocab_size,
        output_dim=32,
        mask_zero=False,
        name='char_embedding'
    )(char_input)
    
    print("Layer 1: Embedding")
    print(f"  Type:           Embedding")
    print(f"  Vocab Size:     {vocab_size}")
    print(f"  Output Dim:     32")
    print(f"  Max Length:     {max_length}")
    print(f"  Output Shape:   (batch, {max_length}, 32)")
    print()
    
    # Layer 2-3: Conv Block 1 (kernel=3, 64 filters)
    x = layers.Conv1D(
        filters=64,
        kernel_size=3,
        padding='same',
        activation='relu',
        kernel_regularizer=regularizers.l2(0.01),
        bias_regularizer=None,
        name='char_conv_block_1'
    )(x)
    
    x = layers.BatchNormalization(name='char_batch_norm_1')(x)
    
    print("Layers 2-3: Conv1D Block 1 + BatchNorm")
    print(f"  Type:              Conv1D + BatchNormalization")
    print(f"  Filters:           64")
    print(f"  Kernel Size:       3")
    print(f"  L2 Regularization: 0.01")
    print(f"  Output Shape:      (batch, {max_length}, 64)")
    print()
    
    # Layer 4-5: Conv Block 2 (kernel=5, 64 filters)
    x = layers.Conv1D(
        filters=64,
        kernel_size=5,
        padding='same',
        activation='relu',
        kernel_regularizer=regularizers.l2(0.01),
        bias_regularizer=None,
        name='char_conv_block_2'
    )(x)
    
    x = layers.BatchNormalization(name='char_batch_norm_2')(x)
    
    print("Layers 4-5: Conv1D Block 2 + BatchNorm")
    print(f"  Type:              Conv1D + BatchNormalization")
    print(f"  Filters:           64")
    print(f"  Kernel Size:       5")
    print(f"  L2 Regularization: 0.01")
    print(f"  Output Shape:      (batch, {max_length}, 64)")
    print()
    
    # Layer 6-7: Conv Block 3 (kernel=7, 64 filters)
    x = layers.Conv1D(
        filters=64,
        kernel_size=7,
        padding='same',
        activation='relu',
        kernel_regularizer=regularizers.l2(0.01),
        bias_regularizer=None,
        name='char_conv_block_3'
    )(x)
    
    x = layers.BatchNormalization(name='char_batch_norm_3')(x)
    
    print("Layers 6-7: Conv1D Block 3 + BatchNorm")
    print(f"  Type:              Conv1D + BatchNormalization")
    print(f"  Filters:           64")
    print(f"  Kernel Size:       7")
    print(f"  L2 Regularization: 0.01")
    print(f"  Output Shape:      (batch, {max_length}, 64)")
    print()
    
    # Layer 8: Global Max Pooling
    x = layers.GlobalMaxPooling1D(name='char_global_maxpool')(x)
    
    print("Layer 8: GlobalMaxPooling1D")
    print(f"  Type:        GlobalMaxPooling1D")
    print(f"  Output Shape: (batch, 64)")
    print()
    
    # Layer 9: Dense 1 (256 units)
    x = layers.Dense(
        units=256,
        activation='relu',
        kernel_regularizer=regularizers.l2(0.01),
        name='char_dense_1'
    )(x)
    
    print("Layer 9: Dense Layer 1")
    print(f"  Type:              Dense")
    print(f"  Units:             256")
    print(f"  Activation:        relu")
    print(f"  L2 Regularization: 0.01")
    print(f"  Output Shape:      (batch, 256)")
    print()
    
    # Layer 10: Dropout (0.5 - HIGHEST among branches)
    x = layers.Dropout(
        rate=0.5,
        name='char_dropout'
    )(x)
    
    print("Layer 10: Dropout")
    print(f"  Type:  Dropout")
    print(f"  Rate:  0.5 (HIGHEST among branches - prevents co-adaptation)")
    print(f"  Output Shape: (batch, 256)")
    print()
    
    # Layer 11: Dense 2 (128 units - branch output)
    x = layers.Dense(
        units=128,
        activation='relu',
        kernel_regularizer=regularizers.l2(0.01),
        name='char_dense_2'
    )(x)
    
    print("Layer 11: Dense Layer 2 (Branch Output)")
    print(f"  Type:              Dense")
    print(f"  Units:             128")
    print(f"  Activation:        relu")
    print(f"  L2 Regularization: 0.01")
    print(f"  Output Shape:      (batch, 128)")
    print()
    
    # Create model
    model = keras.Model(inputs=char_input, outputs=x, name='character_branch')
    
    return model

# Build the model
char_branch_model = build_character_branch(max_length=1024, vocab_size=256)

print("="*80)
print("CHARACTER BRANCH MODEL SUMMARY")
print("="*80)
print()

char_branch_model.summary()

print()
print("="*80)
print("CELL 1 COMPLETE - Character Branch Architecture Built")
print("="*80)
```

```python
# ============================================================================
# PHASE 4 - DAY 43 - CELL 2: BUILD WORD-LEVEL SEMANTIC BRANCH
# ============================================================================
# Goal: Build word branch with word + token type embeddings, Conv1D, pooling
# ============================================================================

print("="*80)
print("DAY 43 - CELL 2: BUILD WORD-LEVEL SEMANTIC BRANCH")
print("="*80)
print()

# Redefine paths
PROJECT_ROOT = Path("MAJOR-PROJECT(SQL)")
PHASE4_ROOT = PROJECT_ROOT / "phase4_architecture"
DAY43_ARTIFACTS = PHASE4_ROOT / "artifacts/day43_word_branch"

print("="*80)
print("WORD-LEVEL BRANCH ARCHITECTURE - SEMANTIC UNDERSTANDING")
print("="*80)
print()

# Build word-level semantic branch
def build_word_branch(max_sequence_length=150, word_vocab_size=5000, token_type_vocab=10):
    """
    Build word-level semantic branch with:
    - Word embeddings (100 dim) + Token type embeddings (16 dim)
    - Concatenation for 116-dim combined representation
    - 1D convolutions (k=3, k=5, 64 filters) for semantic patterns
    - Global average pooling
    - Dense layers for feature transformation
    - Dropout 0.3 for regularization
    - L2 regularization 0.01
    
    Purpose: Extract semantic context from token sequences and types
    """
    
    # Input layers
    word_input = layers.Input(shape=(max_sequence_length,), name='word_input', dtype=tf.int32)
    token_type_input = layers.Input(shape=(max_sequence_length,), name='token_type_input', dtype=tf.int32)
    
    print("Inputs:")
    print(f"  word_input shape: (batch, {max_sequence_length})")
    print(f"  token_type_input shape: (batch, {max_sequence_length})")
    print()
    
    # Layer 1-2: Word Embedding (100 dim)
    word_embedding = layers.Embedding(
        input_dim=word_vocab_size,
        output_dim=100,
        mask_zero=False,
        name='word_embedding'
    )(word_input)
    
    print("Layer 1: Word Embedding")
    print(f"  Vocab Size: {word_vocab_size}")
    print(f"  Output Dim: 100")
    print(f"  Output Shape: (batch, {max_sequence_length}, 100)")
    print()
    
    # Layer 3-4: Token Type Embedding (16 dim)
    token_type_embedding = layers.Embedding(
        input_dim=token_type_vocab,
        output_dim=16,
        mask_zero=False,
        name='token_type_embedding'
    )(token_type_input)
    
    print("Layer 2: Token Type Embedding")
    print(f"  Vocab Size: {token_type_vocab}")
    print(f"  Output Dim: 16")
    print(f"  Output Shape: (batch, {max_sequence_length}, 16)")
    print()
    
    # Layer 5: Concatenate embeddings (100 + 16 = 116)
    concatenated = layers.Concatenate(
        axis=-1,
        name='concat_embeddings'
    )([word_embedding, token_type_embedding])
    
    print("Layer 3: Concatenate Embeddings")
    print(f"  Input 1: (batch, {max_sequence_length}, 100) - word")
    print(f"  Input 2: (batch, {max_sequence_length}, 16) - token type")
    print(f"  Output: (batch, {max_sequence_length}, 116)")
    print()
    
    # Layer 6-7: Conv1D Block 1 (kernel=3, 64 filters)
    x = layers.Conv1D(
        filters=64,
        kernel_size=3,
        padding='same',
        activation='relu',
        kernel_regularizer=regularizers.l2(0.01),
        name='word_conv_block_1'
    )(concatenated)
    
    x = layers.BatchNormalization(name='word_batch_norm_1')(x)
    
    print("Layer 4-5: Conv1D Block 1 + BatchNorm")
    print(f"  Filters: 64")
    print(f"  Kernel Size: 3")
    print(f"  L2 Regularization: 0.01")
    print(f"  Output Shape: (batch, {max_sequence_length}, 64)")
    print()
    
    # Layer 8-9: Conv1D Block 2 (kernel=5, 64 filters)
    x = layers.Conv1D(
        filters=64,
        kernel_size=5,
        padding='same',
        activation='relu',
        kernel_regularizer=regularizers.l2(0.01),
        name='word_conv_block_2'
    )(x)
    
    x = layers.BatchNormalization(name='word_batch_norm_2')(x)
    
    print("Layer 6-7: Conv1D Block 2 + BatchNorm")
    print(f"  Filters: 64")
    print(f"  Kernel Size: 5")
    print(f"  L2 Regularization: 0.01")
    print(f"  Output Shape: (batch, {max_sequence_length}, 64)")
    print()
    
    # Layer 10: Dropout after conv blocks
    x = layers.Dropout(
        rate=0.3,
        name='word_dropout_conv'
    )(x)
    
    print("Layer 8: Dropout (Convolution Block)")
    print(f"  Rate: 0.3 (moderate regularization)")
    print(f"  Output Shape: (batch, {max_sequence_length}, 64)")
    print()
    
    # Layer 11: Global Average Pooling
    x = layers.GlobalAveragePooling1D(name='word_global_avgpool')(x)
    
    print("Layer 9: GlobalAveragePooling1D")
    print(f"  Output Shape: (batch, 64)")
    print(f"  Purpose: Mean of conv features across sequence")
    print()
    
    # Layer 12: Dense 1 (256 units)
    x = layers.Dense(
        units=256,
        activation='relu',
        kernel_regularizer=regularizers.l2(0.01),
        name='word_dense_1'
    )(x)
    
    print("Layer 10: Dense Layer 1")
    print(f"  Units: 256")
    print(f"  L2 Regularization: 0.01")
    print(f"  Output Shape: (batch, 256)")
    print()
    
    # Layer 13: Dropout Dense
    x = layers.Dropout(
        rate=0.3,
        name='word_dropout_dense'
    )(x)
    
    print("Layer 11: Dropout (Dense)")
    print(f"  Rate: 0.3")
    print(f"  Output Shape: (batch, 256)")
    print()
    
    # Layer 14: Dense 2 (128 units - branch output)
    x = layers.Dense(
        units=128,
        activation='relu',
        kernel_regularizer=regularizers.l2(0.01),
        name='word_dense_2'
    )(x)
    
    print("Layer 12: Dense Layer 2 (Branch Output)")
    print(f"  Units: 128")
    print(f"  L2 Regularization: 0.01")
    print(f"  Output Shape: (batch, 128)")
    print()
    
    # Create model
    model = keras.Model(
        inputs=[word_input, token_type_input],
        outputs=x,
        name='word_semantic_branch'
    )
    
    return model

# Build the model
word_branch_model = build_word_branch(
    max_sequence_length=150,
    word_vocab_size=5000,
    token_type_vocab=10
)

print("="*80)
print("WORD-LEVEL BRANCH MODEL SUMMARY")
print("="*80)
print()

word_branch_model.summary()

print()
print("="*80)
print("CELL 2 COMPLETE - Word Branch Architecture Built")
print("="*80)
```

```python
# ============================================================================
# PHASE 4 - DAY 43 - CELL 4: FORWARD PASS VALIDATION (CORRECTED)
# ============================================================================
# Goal: Validate word branch with proper token handling
# ============================================================================

print("="*80)
print("DAY 43 - CELL 4: FORWARD PASS VALIDATION (CORRECTED)")
print("="*80)
print()

# Redefine paths
PROJECT_ROOT = Path("MAJOR-PROJECT(SQL)")
PHASE4_ROOT = PROJECT_ROOT / "phase4_architecture"
DAY43_ARTIFACTS = PHASE4_ROOT / "artifacts/day43_word_branch"

# Reload word training data
print("="*80)
print("LOADING AND PREPARING SAMPLE DATA")
print("="*80)
print()

train_word_path = Path("../notebooks/phase3b_pipeline/data/tokenized/train_word_tokenized_raw.parquet")
word_train_df = pd.read_parquet(train_word_path)

print(f"Loaded {len(word_train_df):,} samples")
print()

# Inspect data structure
print("Data Structure Check:")
print(f"  word_tokens type: {type(word_train_df['word_tokens'].iloc[0])}")
print(f"  token_types type: {type(word_train_df['token_types'].iloc[0])}")
print()

# The issue: word_tokens are STRINGS (not integers)
# We need to convert them to integer indices
# Create vocabulary mapping from unique tokens
print("Creating Token-to-ID Mapping...")

# Get all unique word tokens (this might take a moment)
all_word_tokens = set()
for tokens in word_train_df['word_tokens']:
    if isinstance(tokens, (list, np.ndarray)):
        all_word_tokens.update(tokens)

# Create word vocabulary mapping
word2id = {'<PAD>': 0}
for idx, token in enumerate(sorted(list(all_word_tokens)), 1):
    word2id[str(token)] = idx
    if idx >= 4999:  # Cap at 5000 vocab
        break

print(f"  Created vocabulary with {len(word2id)} unique tokens")
print(f"  Sample tokens: {list(word2id.items())[:5]}")
print()

# Convert tokens to IDs
print("Converting tokens to integer IDs...")

def tokens_to_ids(tokens, vocab_dict, max_id=5000):
    """Convert token strings to integer IDs"""
    ids = []
    for token in tokens:
        token_str = str(token)
        token_id = vocab_dict.get(token_str, 1)  # UNK = 1
        if token_id <= max_id:
            ids.append(token_id)
    return ids

word_tokens_ids = [tokens_to_ids(tokens, word2id) for tokens in word_train_df['word_tokens']]

print(f"  Converted {len(word_tokens_ids):,} sequences")
print(f"  Sample sequence (first 10): {word_tokens_ids[0][:10]}")
print()

# Token types should already be strings, need to map them too
print("Processing Token Types...")

# Get unique token types
unique_token_types = set()
for types_list in word_train_df['token_types']:
    if isinstance(types_list, (list, np.ndarray)):
        unique_token_types.update(types_list)

# Create token type mapping
type2id = {'<PAD>': 0}
for idx, ttype in enumerate(sorted(list(unique_token_types)), 1):
    type2id[str(ttype)] = idx

print(f"  Created token type vocabulary: {len(type2id)} types")
print(f"  Token types: {type2id}")
print()

# Convert token types to IDs
def types_to_ids(types_list, vocab_dict):
    """Convert token type strings to integer IDs"""
    ids = []
    for ttype in types_list:
        type_str = str(ttype)
        type_id = vocab_dict.get(type_str, 0)  # UNK = 0
        ids.append(type_id)
    return ids

token_types_ids = [types_to_ids(types_list, type2id) for types_list in word_train_df['token_types']]

print(f"  Converted {len(token_types_ids):,} token type sequences")
print(f"  Sample types (first 10): {token_types_ids[0][:10]}")
print()

# Now pad sequences to 150
print("Padding sequences to 150...")
max_seq_length = 150

word_tokens_padded = keras.preprocessing.sequence.pad_sequences(
    word_tokens_ids,
    maxlen=max_seq_length,
    padding='post',
    truncating='post',
    value=0,
    dtype='int32'
)

token_types_padded = keras.preprocessing.sequence.pad_sequences(
    token_types_ids,
    maxlen=max_seq_length,
    padding='post',
    truncating='post',
    value=0,
    dtype='int32'
)

labels = np.array(word_train_df['label'].tolist())

print(f"  Word tokens padded shape: {word_tokens_padded.shape}")
print(f"  Token types padded shape: {token_types_padded.shape}")
print()

print("="*80)
print("FORWARD PASS VALIDATION")
print("="*80)
print()

# Create sample batch
sample_size = 64
sample_indices = np.random.choice(len(word_train_df), size=sample_size, replace=False)

sample_word_tokens = word_tokens_padded[sample_indices]
sample_token_types = token_types_padded[sample_indices]
sample_labels = labels[sample_indices]

print(f"Sample batch size: {sample_size}")
print(f"  Word tokens shape: {sample_word_tokens.shape}")
print(f"  Token types shape: {sample_token_types.shape}")
print(f"  Labels: {np.bincount(sample_labels)}")
print()

print("[TESTING] Forward pass through word branch...")
try:
    branch_output = word_branch_model.predict(
        [sample_word_tokens, sample_token_types],
        verbose=0
    )
    
    print("[PASS] Forward pass successful")
    print(f"  Output shape: {branch_output.shape}")
    print(f"  Expected: ({sample_size}, 128)")
    
    if branch_output.shape == (sample_size, 128):
        print("[PASS] Output dimensions correct ✓")
    else:
        print(f"[FAIL] Output dimension mismatch")
    
    print()
    
    # Output statistics
    print("Output Statistics:")
    print(f"  Mean:   {np.mean(branch_output):.6f}")
    print(f"  Std:    {np.std(branch_output):.6f}")
    print(f"  Min:    {np.min(branch_output):.6f}")
    print(f"  Max:    {np.max(branch_output):.6f}")
    print(f"  Zeros:  {np.sum(branch_output == 0)} / {branch_output.size} ({(np.sum(branch_output == 0)/branch_output.size)*100:.2f}%)")
    print()
    
except Exception as e:
    print(f"[FAIL] Forward pass failed: {e}")

print()
print("="*80)
print("VISUALIZATION 1: OUTPUT DISTRIBUTION BY CLASS")
print("="*80)
print()

# Create visualization
fig1 = go.Figure()

benign_indices = np.where(sample_labels == 0)[0]
malicious_indices = np.where(sample_labels == 1)[0]

if len(benign_indices) > 0:
    benign_means = np.mean(branch_output[benign_indices], axis=1)
    fig1.add_trace(go.Box(
        y=benign_means,
        name='Benign (Class 0)',
        marker=dict(color='#4ECDC4'),
        boxmean='sd',
        hovertemplate='<b>Benign Query</b><br>Mean Output: %{y:.4f}<extra></extra>'
    ))

if len(malicious_indices) > 0:
    malicious_means = np.mean(branch_output[malicious_indices], axis=1)
    fig1.add_trace(go.Box(
        y=malicious_means,
        name='Malicious (Class 1)',
        marker=dict(color='#FF6B6B'),
        boxmean='sd',
        hovertemplate='<b>Malicious Query</b><br>Mean Output: %{y:.4f}<extra></extra>'
    ))

fig1.update_layout(
    title='Word Branch Output Distribution by Class<br><sub>Forward Pass on 64 Samples (Benign vs Malicious)</sub>',
    yaxis_title='Mean Output Value (across 128 dimensions)',
    xaxis_title='Query Type',
    height=550,
    width=900,
    hovermode='y',
    showlegend=True,
)

fig1.show()

print()
print("="*80)
print("VISUALIZATION 2: FEATURE ACTIVATION PATTERNS")
print("="*80)
print()

# Visualization: activation patterns
fig2 = go.Figure()

# Calculate statistics per output dimension
dimension_means = np.mean(branch_output, axis=0)
dimension_stds = np.std(branch_output, axis=0)

# Sample every Nth dimension for clarity
sample_rate = max(1, len(dimension_means) // 50)
sampled_dims = np.arange(0, len(dimension_means), sample_rate)

fig2.add_trace(go.Bar(
    x=sampled_dims,
    y=dimension_means[sampled_dims],
    error_y=dict(
        type='data',
        array=dimension_stds[sampled_dims],
        visible=True
    ),
    marker=dict(color=dimension_means[sampled_dims], colorscale='Viridis'),
    name='Output Dimensions',
    hovertemplate='Dim %{x}: Mean=%{y:.4f}<extra></extra>'
))

fig2.update_layout(
    title='Word Branch Output Feature Activation Pattern<br><sub>Mean activation per dimension with error bars (std)</sub>',
    xaxis_title='Output Dimension (sampled)',
    yaxis_title='Mean Activation Value',
    height=550,
    width=900,
    hovermode='x',
    showlegend=False,
)

fig2.show()

print()
print("="*80)
print("SAVING VALIDATION RESULTS")
print("="*80)
print()

# Save validation results
validation_results = pd.DataFrame({
    'Metric': [
        'Total Samples',
        'Batch Size',
        'Output Shape',
        'Mean Output',
        'Std Output',
        'Min Output',
        'Max Output',
        'Benign Samples',
        'Malicious Samples'
    ],
    'Value': [
        len(word_train_df),
        sample_size,
        str(branch_output.shape),
        f"{np.mean(branch_output):.6f}",
        f"{np.std(branch_output):.6f}",
        f"{np.min(branch_output):.6f}",
        f"{np.max(branch_output):.6f}",
        len(benign_indices),
        len(malicious_indices)
    ]
})

validation_csv = DAY43_ARTIFACTS / "forward_pass_validation_results.csv"
validation_results.to_csv(validation_csv, index=False)

print(f"[SAVED] Validation results")
print(f"  Path: {validation_csv}")
print()

# Save vocabulary mappings
vocab_data = pd.DataFrame([
    {'Token ID': vid, 'Token String': vstr} 
    for vstr, vid in sorted(word2id.items(), key=lambda x: x[1])
])

vocab_csv = DAY43_ARTIFACTS / "word_vocabulary_mapping.csv"
vocab_data.head(100).to_csv(vocab_csv, index=False)  # Save first 100 for reference

print(f"[SAVED] Word vocabulary mapping (first 100)")
print(f"  Path: {vocab_csv}")
print()

type_vocab_data = pd.DataFrame([
    {'Token Type ID': vid, 'Token Type': vstr} 
    for vstr, vid in sorted(type2id.items(), key=lambda x: x[1])
])

type_vocab_csv = DAY43_ARTIFACTS / "token_type_vocabulary_mapping.csv"
type_vocab_data.to_csv(type_vocab_csv, index=False)

print(f"[SAVED] Token type vocabulary mapping")
print(f"  Path: {type_vocab_csv}")
print()

print("="*80)
print("CELL 4 COMPLETE - FORWARD PASS VALIDATED")
print("="*80)
print()

print("[SUCCESS] Word Branch Forward Pass Validation Complete")
print("  ✓ Tokens converted to integer IDs")
print("  ✓ Sequences padded to 150")
print("  ✓ Forward pass on 64 samples successful")
print("  ✓ Output shape: (64, 128)")
print("  ✓ Visualizations displayed")
print("  ✓ Vocabulary mappings saved")
print()
```
