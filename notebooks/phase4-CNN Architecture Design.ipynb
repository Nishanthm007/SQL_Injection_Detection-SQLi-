```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, regularizers
import numpy as np

print("="*80)
print("DAY 42: CHARACTER-LEVEL BRANCH REBALANCING")
print("="*80)
print()

# Redefine paths for Day 42
PROJECT_ROOT = Path("MAJOR-PROJECT(SQL)")
PHASE4_ROOT = PROJECT_ROOT / "phase4_architecture"
DAY42_ARTIFACTS = PHASE4_ROOT / "artifacts/day42_char_branch"
DAY42_ARTIFACTS.mkdir(parents=True, exist_ok=True)

print("[INITIALIZING] Day 42 Character Branch Implementation")
print(f"Artifacts Directory: {DAY42_ARTIFACTS}")
print()

# Load character training data
print("="*80)
print("LOADING CHARACTER TRAINING DATA")
print("="*80)
print()

train_char_path = Path("../notebooks/phase3b_pipeline/data/tokenized/train_char_tokenized.parquet")
char_train_df = pd.read_parquet(train_char_path)

print(f"Loaded character training data: {len(char_train_df):,} samples")
print(f"Columns: {list(char_train_df.columns)}")
print(f"Label distribution:")
print(f"  Benign (0):   {(char_train_df['label'] == 0).sum():,}")
print(f"  Malicious (1): {(char_train_df['label'] == 1).sum():,}")
print()

# Analyze character sequence dimensions
max_char_length = char_train_df['char_length'].max()
mean_char_length = char_train_df['char_length'].mean()
median_char_length = char_train_df['char_length'].median()

print(f"Character Sequence Statistics:")
print(f"  Max Length:    {max_char_length}")
print(f"  Mean Length:   {mean_char_length:.2f}")
print(f"  Median Length: {median_char_length:.0f}")
print(f"  Input Dimension: {max_char_length}")
print()

print("="*80)
print("CHARACTER BRANCH ARCHITECTURE - REBALANCED (64 filters)")
print("="*80)
print()

# Build character branch with 64 filters (reduced from 128)
def build_character_branch(max_length=1024, vocab_size=256):
    """
    Build character-level branch with:
    - 64 filters (REDUCED - not 128)
    - 3 hierarchical Conv1D layers (kernels: 3, 5, 7)
    - Batch normalization after each conv layer
    - Dropout 0.5 (highest among branches)
    - L2 regularization 0.01 on conv weights
    
    Purpose: Reduce character branch dominance and control overfitting
    """
    
    char_input = layers.Input(shape=(max_length,), name='char_input', dtype=tf.int32)
    
    # Layer 1: Embedding
    x = layers.Embedding(
        input_dim=vocab_size,
        output_dim=32,
        mask_zero=False,
        name='char_embedding'
    )(char_input)
    
    print("Layer 1: Embedding")
    print(f"  Type:           Embedding")
    print(f"  Vocab Size:     {vocab_size}")
    print(f"  Output Dim:     32")
    print(f"  Max Length:     {max_length}")
    print(f"  Output Shape:   (batch, {max_length}, 32)")
    print()
    
    # Layer 2-3: Conv Block 1 (kernel=3, 64 filters)
    x = layers.Conv1D(
        filters=64,
        kernel_size=3,
        padding='same',
        activation='relu',
        kernel_regularizer=regularizers.l2(0.01),
        bias_regularizer=None,
        name='char_conv_block_1'
    )(x)
    
    x = layers.BatchNormalization(name='char_batch_norm_1')(x)
    
    print("Layers 2-3: Conv1D Block 1 + BatchNorm")
    print(f"  Type:              Conv1D + BatchNormalization")
    print(f"  Filters:           64")
    print(f"  Kernel Size:       3")
    print(f"  L2 Regularization: 0.01")
    print(f"  Output Shape:      (batch, {max_length}, 64)")
    print()
    
    # Layer 4-5: Conv Block 2 (kernel=5, 64 filters)
    x = layers.Conv1D(
        filters=64,
        kernel_size=5,
        padding='same',
        activation='relu',
        kernel_regularizer=regularizers.l2(0.01),
        bias_regularizer=None,
        name='char_conv_block_2'
    )(x)
    
    x = layers.BatchNormalization(name='char_batch_norm_2')(x)
    
    print("Layers 4-5: Conv1D Block 2 + BatchNorm")
    print(f"  Type:              Conv1D + BatchNormalization")
    print(f"  Filters:           64")
    print(f"  Kernel Size:       5")
    print(f"  L2 Regularization: 0.01")
    print(f"  Output Shape:      (batch, {max_length}, 64)")
    print()
    
    # Layer 6-7: Conv Block 3 (kernel=7, 64 filters)
    x = layers.Conv1D(
        filters=64,
        kernel_size=7,
        padding='same',
        activation='relu',
        kernel_regularizer=regularizers.l2(0.01),
        bias_regularizer=None,
        name='char_conv_block_3'
    )(x)
    
    x = layers.BatchNormalization(name='char_batch_norm_3')(x)
    
    print("Layers 6-7: Conv1D Block 3 + BatchNorm")
    print(f"  Type:              Conv1D + BatchNormalization")
    print(f"  Filters:           64")
    print(f"  Kernel Size:       7")
    print(f"  L2 Regularization: 0.01")
    print(f"  Output Shape:      (batch, {max_length}, 64)")
    print()
    
    # Layer 8: Global Max Pooling
    x = layers.GlobalMaxPooling1D(name='char_global_maxpool')(x)
    
    print("Layer 8: GlobalMaxPooling1D")
    print(f"  Type:        GlobalMaxPooling1D")
    print(f"  Output Shape: (batch, 64)")
    print()
    
    # Layer 9: Dense 1 (256 units)
    x = layers.Dense(
        units=256,
        activation='relu',
        kernel_regularizer=regularizers.l2(0.01),
        name='char_dense_1'
    )(x)
    
    print("Layer 9: Dense Layer 1")
    print(f"  Type:              Dense")
    print(f"  Units:             256")
    print(f"  Activation:        relu")
    print(f"  L2 Regularization: 0.01")
    print(f"  Output Shape:      (batch, 256)")
    print()
    
    # Layer 10: Dropout (0.5 - HIGHEST among branches)
    x = layers.Dropout(
        rate=0.5,
        name='char_dropout'
    )(x)
    
    print("Layer 10: Dropout")
    print(f"  Type:  Dropout")
    print(f"  Rate:  0.5 (HIGHEST among branches - prevents co-adaptation)")
    print(f"  Output Shape: (batch, 256)")
    print()
    
    # Layer 11: Dense 2 (128 units - branch output)
    x = layers.Dense(
        units=128,
        activation='relu',
        kernel_regularizer=regularizers.l2(0.01),
        name='char_dense_2'
    )(x)
    
    print("Layer 11: Dense Layer 2 (Branch Output)")
    print(f"  Type:              Dense")
    print(f"  Units:             128")
    print(f"  Activation:        relu")
    print(f"  L2 Regularization: 0.01")
    print(f"  Output Shape:      (batch, 128)")
    print()
    
    # Create model
    model = keras.Model(inputs=char_input, outputs=x, name='character_branch')
    
    return model

# Build the model
char_branch_model = build_character_branch(max_length=1024, vocab_size=256)

print("="*80)
print("CHARACTER BRANCH MODEL SUMMARY")
print("="*80)
print()

char_branch_model.summary()

print()
print("="*80)
print("CELL 1 COMPLETE - Character Branch Architecture Built")
print("="*80)
```
